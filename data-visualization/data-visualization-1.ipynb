# Cleaning & Feature Engineering

```python
df.info()
```

```python
df = df.dropna(subset=['Customer ID'])
```

```python
df.info()
```

```python
df.loc[100:200]
```

```python
# Removing cancellation invoices

df = df[~df['Invoice'].astype(str).str.startswith('C')]
```

```python
df.shape
```

```python
# Removing zero/negative quantities
df = df[df['Quantity'] > 0]
df.shape
```

```python
df = df[df['Price'] > 0]
df.shape
```

```python
df.head(3)
```

```python
df['total_price'] = df['Quantity'] * df['Price']
```

```python
df.info()
```

```python
df['Month'] = df['InvoiceDate'].dt.to_period('M').astype(str)
```

```python
df.head(2)
```

# Data visualization

# What is Matplotlib?

- matplotlib is a powerful 2D plotting library in Python.

- It allows you to create publication-quality figures in a variety of formats (PNG, SVG, PDF, etc.).

- Commonly used module: matplotlib.pyplot, often imported as plt.


# Why Use Matplotlib?

- Core plotting engine behind libraries like Seaborn and Pandas plots.

- Works with NumPy and Pandas natively.

- Gives you low-level control of every visual element (useful for fine-tuning).

```python
import matplotlib.pyplot as plt
```

```python
# get montly revenue
```

```python
df.head()
```

```python
# get montly revenue

# - group by month
# - sum their total_price

monthly_revenue = df.groupby('Month')['total_price'].sum()
```

```python
monthly_revenue
```

```python
# line plot
```

```python
plt.figure(figsize=(12, 6))
plt.plot(monthly_revenue.index, monthly_revenue.values, marker='o')
plt.title('Monthly Revenue Over Time')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.grid(True)
plt.show()
```
