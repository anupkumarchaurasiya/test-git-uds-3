```python
import pandas as pd
```

```python
titanic_dataset_github_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"
```

```python
df = pd.read_csv(titanic_dataset_github_url)
```

```python
df.head(2)
```

```python
df['Embarked'].unique()
```

```python
df['Embarked'] = df['Embarked'].fillna("N")
```

```python
df['Embarked'].unique()
```

```python
df.info()
```

```python
# N, S, C, Q --> (0,1,2,3) ---> Q > N
df['Embarked_cust_cat']= pd.Categorical(
    df['Embarked'],
    categories=['N', 'S', 'C', 'Q'], # Custom order lower to higher (Ascending)
    ordered=True
)
```

```python
df['Embarked_cust_cat'].cat.categories
```

```python
df['Embarked_cust_cat'].cat.codes
```

```python
df['Embarked_cust_cat'] > 'S'
```

```python
df['Embarked_cust_cat_code'] = df['Embarked_cust_cat'].cat.codes
```

# Date Functionality in Pandas

```python
df = pd.DataFrame({
    'event': ['Concert', 'Conference', 'Wedding'],
    'date': ['2025-01-01', '2025-03-15', '2025-07-20']
})
```

```python
df.info()
```

```python
df
```

```python
df['date_pandas'] = pd.to_datetime(df['date'])
# format: optional format (e.g. '%d/%m/%Y')
```

```python
df.info()
```

```python
df.head()
```

```python
df['date_pandas'].dt.year
```

```python
df['date_pandas'].dt.month
```

```python
df['date_pandas'].dt.day
```

```python
df['date_pandas'].dt.day_of_week
```

```python
# 0 - monday
# 6 - sunday
```

```python
df['date_pandas'].dt.day_name()
```

```python
df[df['date_pandas'] > '2025-04-01']
```

```python
df[df['date_pandas'].between('2025-01-01', '2025-06-01')]
```

# Creating Datetime Ranges

Other frequencies:

'M' → month end

'MS' → month start

'W' → weekly

'H', 'T', 'S' → hour, minute, second

```python
pd.date_range(start='2026-02-01', end='2026-02-10', freq='D')
```

```python
pd.date_range(start='2024-01-01', end='2024-05-10', freq='ME')
```

```python
pd.date_range(start='2024-01-01', end='2024-03-10', freq='MS')
```

```python
pd.date_range(start='2024-01-01', end='2024-03-10', freq='W')
```
